I will generate some 200 samples of binary mask and then use HD loss and Procrustes loss

```python
import os
import numpy as np
from PIL import Image, ImageDraw

def generate_random_blob(size=128, num_vertices_range=(5, 10)):
    """
    Generates a random blob mask by drawing a random polygon on a blank mask.
    Returns both the grayscale image (with random intensity inside the blob)
    and its binary mask.
    """
    # Random center
    center_x = np.random.randint(size // 4, 3 * size // 4)
    center_y = np.random.randint(size // 4, 3 * size // 4)
    
    # Choose number of vertices
    num_vertices = np.random.randint(num_vertices_range[0], num_vertices_range[1] + 1)
    
    # Generate random polygon vertices
    angles = np.linspace(0, 2 * np.pi, num_vertices, endpoint=False) + np.random.uniform(0, 2*np.pi/num_vertices, num_vertices)
    min_radius = size // 8
    max_radius = size // 3
    radii = np.random.randint(min_radius, max_radius, num_vertices)
    
    # Create vertex list in (x, y) format
    vertices = [
        (
            center_x + int(radii[i] * np.cos(angles[i])),
            center_y + int(radii[i] * np.sin(angles[i]))
        )
        for i in range(num_vertices)
    ]
    
    # Create blank mask and draw polygon
    mask_img = Image.new('L', (size, size), 0)
    draw = ImageDraw.Draw(mask_img)
    draw.polygon(vertices, outline=1, fill=1)
    mask = np.array(mask_img, dtype=np.uint8)
    
    # Create grayscale image: background = 0, blob area = random intensity
    image = np.zeros((size, size), dtype=np.uint8)
    intensity = np.random.randint(100, 256)
    image[mask == 1] = intensity
    
    return image, mask

# Directory to save images
output_dir = "/home/sbk29/data/github_AR/AR_detection/synthetic_data/"
os.makedirs(output_dir, exist_ok=True)

# Generate and save 10 samples
num_samples = 500
for i in range(1, num_samples + 1):
    img, mask = generate_random_blob(size=128)
    
    # Save grayscale image
    img_pil = Image.fromarray(img)
    img_filename = os.path.join(output_dir, f"image_{i:03d}.png")
    img_pil.save(img_filename)
    
    # Save binary mask (scale mask values to 0/255 for visualization)
    mask_pil = Image.fromarray(mask * 255)
    mask_filename = os.path.join(output_dir, f"mask_{i:03d}.png")
    mask_pil.save(mask_filename)

output_dir
```
